```python
job = 'Data Analyst'
salary= 100000
location = 'india'
```

```python
class job_info:
  def __init__(self, job, salary, location):
    self.job = job
    self.salary = salary
    self.location = location

  def display(self):
    print(f"Job: {self.job}")
    print(f"Salary: {self.salary}")
    print(f"Location: {self.location}")
```

```python
job_1 = job_info(job, salary, location)
```

```python
job_1.display()
```
