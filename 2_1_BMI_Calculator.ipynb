```python
name = input("Whats Your name?")
Weight = int(input("What's your weight in pound: "))
Height = int(input("What's your height in inches:"))

BMI = (Weight * 703)/(Height * Height)

print("Your BML is: ", BMI)

if BMI > 0:
    if(BMI < 16):
        print(name +", You are severely underweight")
    elif(BMI <= 18.5):
        print(name +", You are underweight")
    elif(BMI <= 25):
        print(name +", You are Normal")
    elif(BMI <=30):
        print(name +", You are Overweight")
    elif(BMI <= 35):
        print(name +", You are Obese class 1")
    elif(BMI <= 40):
        print(name +", You are Obese class 2")
    else:
        print(name +", You are Obese class 3")
```

Classification	BMI range - kg/m2

Severe Thinness	< 16

Moderate Thinness	16 - 17

Mild Thinness	17 - 18.5

Normal	18.5 - 25

Overweight	25 - 30

Obese Class I	30 - 35

Obese Class II	35 - 40

Obese Class III	> 40
