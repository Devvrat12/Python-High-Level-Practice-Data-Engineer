```python
flavor = ['Vanilla', 'Choclate', 'Butterscotch']
toppings = ['Hot Fudge', 'Oreos', 'Marshmallows']
```

```python
for one in flavor:
  for two in toppings:
    print(one, "topped with", two)
```
