```python
import os, shutil
```

```python
path = r"E:/stock videos/"
```

```python
file_name = os.listdir(path)
```

```python
file_name
```

```python
folder_names = ['mp4 files', 'sql files', 'png files']

for loop in range(0, 3):
    if not os.path.exists(path + folder_names[loop]):
        os.makedirs(path + folder_names[loop])

for file in file_name:
    if ".mp4" in file and not os.path.exists(path + "mp4 files/"+ file):
        shutil.move(path + file, path + "mp4 files/"+ file)
    elif ".png" in file and not os.path.exists(path + "png files/"+ file):
        shutil.move(path + file, path + "png files/"+ file)
    elif ".sql" in file and not os.path.exists(path + "sql files/"+ file):
        shutil.move(path + file, path + "sql files/"+ file)
```
